```python
import pandas as pd

# Raw URL for the CSV file on GitHub
url = 'https://raw.githubusercontent.com/HumanCompatibleAI/PARETO/main/survey_data/likert-responses.csv'

try:
    df = pd.read_csv(url)
    print("Data downloaded successfully!")
    display(df.head())
except Exception as e:
    print(f"An error occurred while downloading the data: {e}")
```

```python
# 1. Total number of participants
total_participants = df['respondent_id'].nunique()

# Group by respondent to find unique stances
respondent_stances = df.groupby('respondent_id')['user_stance_side'].unique()

# 2. Number of respondents with multiple stance sides
multiple_stances_mask = respondent_stances.apply(len) > 1
num_multiple_stances = multiple_stances_mask.sum()

# 3. Number of respondents with only one stance side
single_stance_respondents = respondent_stances[~multiple_stances_mask]
num_single_stance = len(single_stance_respondents)

# Extract the single stance value for breakdown
single_stance_values = single_stance_respondents.apply(lambda x: x[0])
breakdown = single_stance_values.value_counts()

# Output results
print(f"Total number of participants: {total_participants}")
print(f"Number of respondents with multiple stance sides: {num_multiple_stances}")
print(f"Number of respondents with only one stance side: {num_single_stance}")
print("  - Conservative: ", breakdown.get('conservative', 0))
print("  - Liberal:      ", breakdown.get('liberal', 0))
```

```python
df_likert_2 = df[df['likert'] == 'likert_2']
df_likert_5 = df[df['likert'] == 'likert_5']

print(f"df_likert_2 shape: {df_likert_2.shape}")
print(f"df_likert_5 shape: {df_likert_5.shape}")

print("\nFirst 3 rows of df_likert_2:")
display(df_likert_2.head(3))

print("\nFirst 3 rows of df_likert_5:")
display(df_likert_5.head(3))
```

```python
# Compare qid and question_id columns
print("Sample of qid and question_id side by side:")
display(df[['qid', 'question_id']].drop_duplicates().head(10))

print("\nUnique qid count:", df['qid'].nunique())
print("Unique question_id count:", df['question_id'].nunique())
```

```python
import pandas as pd

# Filter df_likert_2 for only liberal and conservative stances to simplify calculation
df_filtered = df_likert_2[df_likert_2['user_stance_side'].isin(['liberal', 'conservative'])]

# Group by qid and user_stance_side, aggregating count and mean of user_score
stance_stats = df_filtered.groupby(['qid', 'user_stance_side']).agg(
    count=('user_stance_side', 'size'),
    avg_user_score=('user_score', 'mean')
).unstack(fill_value=0)

# Flatten multi-level columns for cleaner presentation
stance_stats.columns = [f"{col[1]}_{col[0]}" for col in stance_stats.columns]
stance_stats = stance_stats.reset_index()

# Display the aggregated statistics
display(stance_stats.head())
```

```python
# Filter df_likert_5 for only liberal and conservative stances
df_filtered_5 = df_likert_5[df_likert_5['user_stance_side'].isin(['liberal', 'conservative'])]

# Group by qid and user_stance_side, aggregating count and mean of user_score
stance_stats_5 = df_filtered_5.groupby(['qid', 'user_stance_side']).agg(
    count=('user_stance_side', 'size'),
    avg_user_score=('user_score', 'mean')
).unstack(fill_value=0)

# Flatten multi-level columns for cleaner presentation
stance_stats_5.columns = [f"{col[1]}_{col[0]}" for col in stance_stats_5.columns]
stance_stats_5 = stance_stats_5.reset_index()

# Display the aggregated statistics for df_likert_5
display(stance_stats_5.head())
```

```python
# Extract unique metadata columns per qid from the original DataFrame
metadata_df = df[['qid', 'issue', 'model', 'model_stance']].drop_duplicates()

# Merge the two DataFrames on qid
combined_stats = pd.merge(
    stance_stats[['qid', 'conservative_avg_user_score', 'liberal_avg_user_score']],
    stance_stats_5[['qid', 'conservative_avg_user_score', 'liberal_avg_user_score']],
    on='qid',
    suffixes=('_stats', '_stats_5')
)

# Rename the columns as specified
combined_stats = combined_stats.rename(columns={
    'conservative_avg_user_score_stats': 'c_approval',
    'liberal_avg_user_score_stats': 'l_approval',
    'conservative_avg_user_score_stats_5': 'c_view',
    'liberal_avg_user_score_stats_5': 'l_view'
})

# Round the approval and view scores to 2 decimal places
score_cols = ['c_approval', 'l_approval', 'c_view', 'l_view']
combined_stats[score_cols] = combined_stats[score_cols].round(2)

# Merge the metadata (issue, model, model_stance) back on qid
combined_stats = pd.merge(metadata_df, combined_stats, on='qid', how='inner')

# Display the combined DataFrame
display(combined_stats.head(100))
```

```python
## Download model response jsons

import requests
import json

# Query the GitHub API to list files in the stimuli_json directory
api_url = "https://api.github.com/repos/HumanCompatibleAI/PARETO/contents/stimuli_json"
response = requests.get(api_url)

if response.status_code == 200:
    contents = response.json()
    file_names = [item['name'] for item in contents if item['type'] == 'file']
    print("Files found in 'stimuli_json':")
    for name in file_names:
        print(f" - {name}")

    # Download all JSON files found there
    stimuli_data = {}
    raw_base_url = "https://raw.githubusercontent.com/HumanCompatibleAI/PARETO/main/stimuli_json/"
    for name in file_names:
        if name.endswith('.json'):
            file_url = f"{raw_base_url}{name}"
            file_response = requests.get(file_url)
            if file_response.status_code == 200:
                stimuli_data[name.replace('.json', '')] = file_response.json()
                print(f"Successfully downloaded {name}!")
            else:
                print(f"Failed to download {name}: Status {file_response.status_code}")
else:
    print(f"Failed to query directory contents: Status {response.status_code}")
    print(response.text)
```

```python
def get_stimulus_details(row):
    model = row['model']
    stance = row['model_stance']
    qid = row['qid']

    # Extract the unique prompt hash (the last part of the qid)
    qid_suffix = qid.split('-')[-1]

    # Map model and stance to the correct key in stimuli_data
    key = None
    if model == 'claude':
        key = 'anthropic'
    elif model == 'gemini':
        key = 'gemini'
    elif model == 'grok':
        key = 'grok'
    elif model == 'llama':
        key = 'llama'
    elif model == 'gpt':
        if stance == 'default':
            key = 'openai_default'
        elif stance == 'balanced':
            key = 'openai_balanced'
        elif stance == 'against':
            key = 'openai_oppose'
        elif stance == 'support':
            key = 'openai_support'
        elif stance == 'chunked':
            key = 'openai_default'

    if key and key in stimuli_data:
        data = stimuli_data[key]
        if qid_suffix in data:
            val = data[qid_suffix]
            # Extract first item from list structure
            if isinstance(val, list) and len(val) > 0:
                val = val[0]

            if isinstance(val, dict):
                question = val.get('question', None)
                answer = val.get('answer', None)
                return pd.Series([question, answer])

    return pd.Series([None, None])

# Apply the mapping function to add both 'original_question' and 'model_response'
combined_stats[['model_prompt', 'response']] = combined_stats.apply(get_stimulus_details, axis=1)

# Reorder the columns so that the rating metrics are on the extreme right
ordered_columns = [
    'qid', 'issue', 'model', 'model_stance',
    'model_prompt', 'response',
    'c_approval', 'l_approval', 'c_view', 'l_view'
]
combined_stats = combined_stats[ordered_columns]

# Display the updated DataFrame with the new column layout
pd.set_option('display.max_colwidth', 150)
display(combined_stats.head())
```
